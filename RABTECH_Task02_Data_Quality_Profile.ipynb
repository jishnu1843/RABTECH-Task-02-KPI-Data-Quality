## 6. Data Quality Findings Summary

The profiling identified the following data quality issues:

1. Missing values in order_date, city, and discount_pct.
2. One duplicate order record and one duplicated order_id (RT-1004).
3. Invalid quantity values: -1 and "two".
4. Invalid discount value: 105%.
5. Inconsistent categorical values: "paid" and "student".
6. One invalid order date: 2026-13-10.
7. order_date is stored as an object/string and requires date standardization.
8. Valid order dates range from 03-Jan-2026 to 18-Jan-2026.

No data was modified during profiling. Data cleaning and correction will be performed in Task 03.
